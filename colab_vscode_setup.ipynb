```python
# Start code-server in background
import subprocess
import time

# Start code-server
process = subprocess.Popen(
    ['code-server', '--bind-addr', '127.0.0.1:8080', '--auth', 'password'],
    stdout=subprocess.PIPE,
    stderr=subprocess.PIPE
)

print("Starting code-server...")
time.sleep(5)
print("code-server started!")
```

```python
# Create tunnel with localtunnel
import subprocess
import time

print("🌐 Creating tunnel...")

# Start localtunnel in background
tunnel_process = subprocess.Popen(
    ['lt', '--port', '8080', '--subdomain', 'xuannhi-vscode'],
    stdout=subprocess.PIPE,
    stderr=subprocess.PIPE,
    text=True
)

# Wait for tunnel to establish
time.sleep(5)

# Try to get the URL
try:
    # Check if tunnel is running
    result = subprocess.run(['ps', 'aux'], capture_output=True, text=True)
    if 'lt' in result.stdout:
        print("✅ Tunnel created!")
        print("\n🔗 Your VS Code URL:")
        print("   👉 https://xuannhi-vscode.loca.lt")
        print("\n📌 Open this URL in a new browser tab")
    else:
        print("⚠️ Tunnel may not have started")
except:
    print("⚠️ Could not verify tunnel status")
    print("Expected URL: https://xuannhi-vscode.loca.lt")
```
