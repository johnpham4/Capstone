```python
# Xem ket qua
from pathlib import Path
out_dir = Path('dataset/outsrc')
pngs = sorted(out_dir.glob('*.png'))
print('Generated images:', len(pngs))
print('Sample files:')
for p in pngs[:10]:
    print(p)

failed_path = out_dir / 'failed_problems.json'
print('Has failed_problems.json:', failed_path.exists())
```

```python
# Hien thi nhanh 1 anh dau ra
import matplotlib.pyplot as plt
from PIL import Image

if pngs:
    img = Image.open(pngs[0])
    plt.figure(figsize=(8, 8))
    plt.imshow(img)
    plt.axis('off')
    plt.title(pngs[0].name)
    plt.show()
else:
    print('Khong co anh nao duoc tao.')
```
